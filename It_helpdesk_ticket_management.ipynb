# IT Help Desk Ticket Management Database

## Project Scenario

For this assignment, I will create a simple IT help desk database using TinyDB. The database will keep track of users and the IT support tickets they submit.

A help desk may receive issues such as password resets, VPN problems, software errors, and MFA issues. This database will make it easier to organize the tickets and search for specific types of problems.

## Database Design

I will use two tables in my database: users and tickets.

### Users Table

The users table will contain:

- user_id
- name
- department

### Tickets Table

The tickets table will contain:

- ticket_id
- user_id
- issue_type
- priority
- status

The user_id shows which user submitted the ticket. Since one user can submit several tickets, this is a one-to-many relationship.

## Setting Up TinyDB

First, I need to install and import TinyDB. TinyDB stores the data in a JSON file. I will also import Query, which will let me search for specific information in the database later.

```python
# Import TinyDB
from tinydb import TinyDB, Query

print("TinyDB imported successfully.")
```

## Creating the Database

Now I will create the help desk database. I will use one table for users and another table for support tickets.

```python
# Create the database
db = TinyDB('helpdesk_db.json')

# Create the tables
users = db.table('users')
tickets = db.table('tickets')

print("Help desk database created.")
```

## Adding Initial Data

Next I will add some sample users to the database. I am keeping the same fields for each user so the records stay consistent.

```python
# Clear old records before adding the sample data
users.truncate()
tickets.truncate()

# Add sample users
users.insert_multiple([
    {'user_id': 101, 'name': 'Claudia Galdamez', 'department': 'Information Technology'},
    {'user_id': 102, 'name': 'Nathan Miller', 'department': 'Finance'},
    {'user_id': 103, 'name': 'Tanya Brown', 'department': 'Student Services'},
    {'user_id': 104, 'name': 'Maria Lopez', 'department': 'Human Resources'},
    {'user_id': 105, 'name': 'James Smith', 'department': 'Admissions'}
])

print("Users added successfully.")
```

### Adding Support Tickets

Now I will add some sample IT support tickets with different issue types, priorities, and statuses so I can use them later for searches and reports.

```python
# Add sample IT support tickets
tickets.insert_multiple([
    {'ticket_id': 1001, 'user_id': 101, 'issue_type': 'Password Reset',
     'priority': 'Medium', 'status': 'Closed'},

    {'ticket_id': 1002, 'user_id': 102, 'issue_type': 'VPN',
     'priority': 'High', 'status': 'Open'},

    {'ticket_id': 1003, 'user_id': 103, 'issue_type': 'Software',
     'priority': 'Low', 'status': 'Open'},

    {'ticket_id': 1004, 'user_id': 101, 'issue_type': 'MFA',
     'priority': 'High', 'status': 'Open'},

    {'ticket_id': 1005, 'user_id': 104, 'issue_type': 'Password Reset',
     'priority': 'Medium', 'status': 'Closed'},

    {'ticket_id': 1006, 'user_id': 105, 'issue_type': 'Software',
     'priority': 'High', 'status': 'Open'},

    {'ticket_id': 1007, 'user_id': 102, 'issue_type': 'MFA',
     'priority': 'Medium', 'status': 'Closed'},

    {'ticket_id': 1008, 'user_id': 103, 'issue_type': 'VPN',
     'priority': 'High', 'status': 'Open'}
])

print("Tickets added successfully.")
```

# CRUD Operations

CRUD stands for Create, Read, Update, and Delete. I will test each operation using the help desk database.

## Create

For the Create operation, I will add a new support ticket to the database. This represents a user submitting a new issue to the help desk.

```python
# Create a new support ticket
new_ticket = {
    'ticket_id': 1009,
    'user_id': 104,
    'issue_type': 'Email',
    'priority': 'Medium',
    'status': 'Open'
}

tickets.insert(new_ticket)

print("New ticket added:")
print(new_ticket)
```

## Read

The Read operation is used to find information that is already stored in the database. I will search for ticket 1009 to make sure the new ticket was saved.

```python
# Create a query object
Ticket = Query()

# Search for ticket 1009
result = tickets.search(Ticket.ticket_id == 1009)

print("Search result:")
for ticket in result:
    print(ticket)
```

## Update

The Update operation is used when information in an existing record needs to change. For this example, I will change ticket 1009 from Open to Closed to show that the issue was resolved.

```python
# Update ticket 1009 from Open to Closed
tickets.update(
    {'status': 'Closed'},
    Ticket.ticket_id == 1009
)

# Display the updated ticket
updated_ticket = tickets.search(Ticket.ticket_id == 1009)

print("Updated ticket:")
for ticket in updated_ticket:
    print(ticket)
```

## Delete

The Delete operation removes a record from the database. I will delete ticket 1009 since it was only added to demonstrate the CRUD operations.

```python
# Delete ticket 1009
tickets.remove(Ticket.ticket_id == 1009)

# Check if the ticket still exists
result = tickets.search(Ticket.ticket_id == 1009)

print("Search after deletion:")
print(result)
```

The CRUD operations helped me see how TinyDB works with records. I created a ticket, searched for it, updated the status, and then deleted it. This is pretty similar to a help desk system where tickets are added and updated as issues are worked on.

# Complex Queries and Data Relationships

Next, I will use queries with more than one condition to search through the ticket data. This makes it easier to quickly find specific tickets in a help desk system.

## Finding High Priority Open Tickets

First I want to find tickets that are both high priority and still open to see which tickets need to be prioritized first.

```python
# Find tickets that are High priority AND Open
high_open_tickets = tickets.search(
    (Ticket.priority == 'High') &
    (Ticket.status == 'Open')
)

print("High Priority Open Tickets:")

for ticket in high_open_tickets:
    print(ticket)
```

The query found four tickets that were both High priority and still Open. This could be useful in a help desk because it makes it easier to find the tickets that may need attention first.

## Searching Multiple Issue Types

Next, I will search for tickets related to either VPN or MFA issues. This helps group similar access problems together.

```python
# Find tickets that are VPN OR MFA issues
access_issues = tickets.search(
    (Ticket.issue_type == 'VPN') |
    (Ticket.issue_type == 'MFA')
)

print("VPN and MFA Tickets:")

for ticket in access_issues:
    print(ticket)
```

This query returned tickets that were either VPN or MFA issues. Using OR made it possible to search for both types of issues at the same time instead of running two separate searches.

## One-to-Many Relationship

The user_id connects each user to their support tickets. Since one user can have multiple tickets, this creates a one-to-many relationship.

For this example, I will look at all the tickets for Claudia Galdamez, who has user_id 101.

```python
# Search for Claudia's user record
User = Query()

claudia = users.search(User.user_id == 101)

print("User:")
for user in claudia:
    print(user)

# Find all tickets connected to Claudia
claudia_tickets = tickets.search(Ticket.user_id == 101)

print("\nTickets submitted by Claudia Galdamez:")
for ticket in claudia_tickets:
    print(ticket)
```

The results show that Claudia Galdamez is connected to two support tickets through user_id 101.

# Data Integrity

Data integrity means keeping the information in the database correct. For example, each ticket should have its own ticket_id. If two tickets have the same ID, it could cause confusion when searching or updating a ticket.

To prevent this, I will add a check to make sure the ticket_id does not already exist before adding a new ticket.

```python
# Check for a duplicate ticket ID
duplicate_ticket = {
    'ticket_id': 1002,
    'user_id': 105,
    'issue_type': 'Email',
    'priority': 'Low',
    'status': 'Open'
}

# Only add the ticket if the ticket_id does not already exist
existing_ticket = tickets.search(
    Ticket.ticket_id == duplicate_ticket['ticket_id']
)

if existing_ticket:
    print("Ticket ID already exists. Duplicate was not added.")
else:
    tickets.insert(duplicate_ticket)
    print("Ticket added successfully.")
```

The database found that ticket 1002 already existed, so the duplicate was not added. This helps prevent duplicate tickets and keeps the data more consistent.

# Performance Evaluation

Next, I will measure how long it takes TinyDB to search the ticket database. Since this database only has a small number of records, the search should be very fast.

```python
# Import time to measure the query speed
import time

# Start the timer
start_time = time.perf_counter()

# Search for all open tickets
open_tickets = tickets.search(Ticket.status == 'Open')

# Stop the timer
end_time = time.perf_counter()

# Calculate the search time
search_time = end_time - start_time

print("Open tickets found:", len(open_tickets))
print("Search time:", search_time, "seconds")
```

The query found 5 open tickets and took about 0.0007 seconds to run. Since this database only has a small number of records, TinyDB was able to search it very quickly.

# Data Aggregation and Reporting

Now I will summarize the tickets by issue type. This will show how many tickets there are for each type of IT problem.

```python
# Count the number of tickets for each issue type
issue_counts = {}

for ticket in tickets.all():
    issue = ticket['issue_type']

    if issue in issue_counts:
        issue_counts[issue] += 1
    else:
        issue_counts[issue] = 1

print("Tickets by Issue Type:")

for issue, count in issue_counts.items():
    print(issue, ":", count)
```

The report shows 2 tickets for each issue type: Password Reset, VPN, Software, and MFA. No single issue type appears more often than the others in this sample database.

## Ticket Status Report

I will group the tickets by status. This will show how many tickets are Open and how many are Closed.

```python
# Count tickets by status
status_counts = {}

for ticket in tickets.all():
    status = ticket['status']

    if status in status_counts:
        status_counts[status] += 1
    else:
        status_counts[status] = 1

print("Tickets by Status:")

for status, count in status_counts.items():
    print(status, ":", count)
```

The report shows that 5 tickets are still Open and 3 are Closed. This gives a summary of how many support issues are still being worked on compared to how many have been completed.
